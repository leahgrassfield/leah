```python
#5 Lets Play Checkers!
```

```python
board = np.zeros((8,8))
board
```

```python
board
```

```python
def checkerboard1():
    board[0::2,::2] = 1
    return board
print(checkerboard1())
```

```python
def checkerboard():
    board[1::2, 1::2] = 1
    return board
print(checkerboard())
```

```python
checkerboard()
reverse_checkerboard = checkerboard()[:,::-1]
reverse_checkerboard
```

```python
#6 The Expanding Universe
```

```python
universe = np.array(['galaxy', 'clusters'])
```

```python
def expansion(str_array, spaces):
    for i in range(len(str_array)):
        str_array[i] = str_array[i] + ' '*spaces
    return str_array
expansion(universe, 2)
```
