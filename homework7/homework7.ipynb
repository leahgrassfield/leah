```python
import matplotlib.pyplot as plt
```

```python
import seaborn as sns
```

```python
penguin = sns.load_dataset("penguins")
penguin
```

```python
penguin1 = penguin.dropna()
penguins = (penguin1)
```

```python
#species = species 
#Island = island they live 
#bill_length_mm = bill length in mm
#bill_depth_m = bill depth in mm
#flipper_length_mm = flipper length in mm
#body_mass_g = body mass in grams
#sex = male or female
```

```python
fig, axs = plt.subplots(2, 2)
```

```python
bill_length_mm = penguins.iloc[:,2]
bill_depth_mm = penguins.iloc[:,3]
flipper_length_mm = penguins.iloc[:,4]
body_mass_g = penguins.iloc[:,5]
species = penguins.iloc[:,1]
```

```python
fig, axs = plt.subplots(2, 2)
axs[0, 0].scatter(bill_length_mm, bill_depth_mm)
axs[0, 1].scatter(flipper_length_mm, body_mass_g)
sns.histplot(x=species, ax=axs[1, 0])
sns.boxplot(x=body_mass_g, ax=axs[1, 1])
plt.show()
```

```python
fig, axs = plt.subplots(2, 2)
axs[0, 0].scatter(bill_length_mm, bill_depth_mm, c='purple')
axs[0, 0].set_title('Bill Length vs. Bill Depth')
axs[0, 0].set_xlabel('Bill Length')
axs[0, 0].set_ylabel('Bill Depth')
axs[0, 1].scatter(flipper_length_mm, body_mass_g, c='green')
axs[0, 1].set_title('Flipper Length vs. Body Mass')
axs[0, 1].set_xlabel('Flipper Length')
axs[0, 1].set_ylabel('Body Mass')
sns.histplot(x=species, ax=axs[1, 0], color='teal')
axs[1,0].set_title('Penguins per Island')
axs[1, 0].set_xlabel('Islands')
axs[1, 0].set_ylabel('Penguins')
sns.boxplot(x=body_mass_g, ax=axs[1, 1], color = 'pink')
axs[1, 1].set_title('Penguin Body Mass')
axs[1, 1].set_xlabel('Body Mass')
plt.tight_layout()
plt.show()
```

```python
#Purple is bill length vs depth, the correlation between the two
#green is correlation between flipper length and body mass
#teal is the amount of penguins at each island
#pink is a plot of the penguin body mass
#I learned that the dataset has a good spread of the Penguins across the three islands, but is definitely limited by the fact that it is only those islands
```
