```python
import numpy as np
```

```python
import matplotlib.pyplot as plt
```

```python
data = np.loadtxt("../homework8/star_data.txt", skiprows=1)
data
```

```python
data.shape
```

```python
#There are 2642 rows
```

```python
magnitude = data[:,0]
temperature = data [:,1]
```

```python
np.argmin(magnitude)
#corresponds to star 1988
```

```python
np.argmax(magnitude)
#corresponds to star 264
```

```python
np.argmin(temperature)
#star 1923
```

```python
np.argmax(temperature)
#star 2613
```

```python
#The brightest star is number 1988, and the dimmest is number 264
```

```python
log_magnitude = np.log(magnitude)
log_temperature = np.log(temperature)
plt.figure(figsize = (8,6))
plt.scatter(log_temperature, log_magnitude)
plt.gca().invert_xaxis()
plt.gca().invert_yaxis()
plt.show()
```

```python
log_magnitude = np.log(magnitude)
log_temperature = np.log(temperature)
plt.figure(figsize = (8,6))
plt.scatter(log_temperature, log_magnitude, label = 'stars')
plt.gca().invert_xaxis()
plt.gca().invert_yaxis()
#brightest
plt.scatter(log_temperature[1988], log_magnitude[1988], color='black', marker = '*', label = 'brightest')

#dimmest
plt.scatter(log_temperature[264], log_magnitude[264], color='purple', marker = '*', label = 'dimmest')

#hottest
plt.scatter(log_temperature[2613], log_magnitude[2613], color='green',marker = '*',label = 'hottest')

#coldest
plt.scatter(log_temperature[1923], log_magnitude[1923], color='orange',marker = '*', label = 'coldest')

plt.title('Hertzsprung-Russel Diagram')
plt.xlabel('Temperature (log K)')
plt.ylabel('Magnitude')

plt.legend()
plt.show()
```
