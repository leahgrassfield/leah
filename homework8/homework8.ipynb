```python
import numpy as np
```

```python
import matplotlib.pyplot as plt
```

```python
def make_sin_wave(x, A, w):
    return A * np.sin(w*x)
```

```python
x = np.linspace(0,2*(np.pi), num=1000)
```

```python
amplitude = np.array([.5,1, 1.5, 2, 2.5])
frequency = np.array([1, 2, 3, 4, 5])
```

```python
fig, axes = plt.subplots(1,3, figsize=(25,5))
for A, w in zip(amplitude, frequency):
    axes[0].plot(x,make_sin_wave(x, A, 1), label = A)
    axes[1].plot(x,make_sin_wave(x, 1, w), label = w)
    axes[2].plot(x,make_sin_wave(x, A, w), label = w)
axes[0].set_title('Varied Amplitude')
axes[0].set_xlabel('x value')
axes[0].set_ylabel('y value')
axes[0].legend()

axes[1].set_title('Varied Frequency')
axes[1].set_xlabel('x value')
axes[1].set_ylabel('y value')
axes[1].legend()

axes[2].set_title('Varied Frequency and Amplitude')
axes[2].set_xlabel('x value')
axes[2].set_ylabel('y value')
axes[2].legend()
plt.show()
```
